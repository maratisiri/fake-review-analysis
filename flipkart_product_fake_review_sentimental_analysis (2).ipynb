**IMPORT LIBRARIES**:

```python
# 1: lets Imports required libraies for this project
import pandas as pd         # pandas  used to work with tables
import numpy as np          #numpy  used for numerical operations
import re  # clean text using patterns
import string # remove punctuation

import seaborn as sns #draw graphs
import matplotlib.pyplot as plt   #plotings

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC #works only with linear decision boundary, #is faster,optimized version of svm that works with linear decision boundaries
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix #modelperformance, #overal correctness,#detailed per-class performance,#breakdown of predit vs actual values

```

# **LOAD THE DATASET**:

```python
#2:lets Load the raw datasets
df_flipkart_raw = pd.read_csv('flipkart.csv', encoding='utf-8', on_bad_lines='skip')   # ensures text is read correctly -- utf-8
df_amazon_raw  = pd.read_csv('amazon.csv',   encoding='utf-8', on_bad_lines='skip')  ## skips broken rows instead of crashing -- bad lines

display("Flipkart columns:\n", df_flipkart_raw.columns)
display("\nAmazon columns:\n", df_amazon_raw.columns)
```

#lets clean the flipkart dataset

```python
#  3: Cleaning the Flipkart dataset
df_flipkart_data = df_flipkart_raw.rename(columns={
    'product_name': 'category',  # product category

    'Review': 'review_text',  # review sentence
    'Rate': 'rating',        # star rating
    'Sentiment': 'label'      # sentiment label
})

# Mapthe text sentiment to numeric labels from flipkart data
df_flipkart_data['label'] = df_flipkart_data['label'].astype(str).str.strip().str.lower().map({     #everything must be text,#removes punct,#changes to lower cases
    'positive': 1,
    'negative': 0,
    'neutral': 2
})

# Drop rows with missing review or label from flipkart data
df_flipkart_data = df_flipkart_data.dropna(subset=['review_text', 'label']) #removes rows where review_text,label are missing

# lets set a platform column as flipkart
df_flipkart_data['platform'] = 'Flipkart'

display(df_flipkart_data[['review_text','rating','label','platform']].head()) #display first 5 rows
print("Flipkart shape:", df_flipkart_data.shape)
```

#lets clean amazon dataset

```python


#4: lets Clean Amazon dataset
df_amazon_data = df_amazon_raw.rename(columns={
    'ProductId': 'category',
    'Text': 'review_text',
    'Score': 'rating'
})

# Drop the  missing values of  rating or review_text
df_amazon_data = df_amazon_data.dropna(subset=['rating', 'review_text']) #removes rows where review_text,rating are missing

# make sure rating must be numeric
df_amazon_data['rating'] = pd.to_numeric(df_amazon_data['rating'], errors='coerce') #Converts values into numbers,#If conversion fails/ replace value with NaN
df_amazon_data = df_amazon_data.dropna(subset=['rating'])

# Map the rating (1-5) to sentiment label: 0=neg, 1=pos, 2=neutral
def rating_to_label(score):
    if score <= 2:
        return 0      # negative sentiment
    elif score == 3:-+
        return 2      # neutral sentiment
    else:
        return 1      # positive sentiment

df_amazon_data['label'] = df_amazon_data['rating'].apply(rating_to_label) #Applies the function to each row

df_amazon_data['platform'] = 'Amazon'

display(df_amazon_data[['review_text','rating','label','platform']].head()) #shows first five rows
print("Amazon shape:", df_amazon_data.shape)
```

#text cleaning

```python
#  5: Text cleaning function
def clean_text(text):
    text = str(text).lower()
    # remove HTML tags like <br>, <br/>, <...>
    text = re.sub(r'<br\s*/?>', ' ', text)
    text = re.sub(r'<.*?>', ' ', text)
    # remove punctuation
    text = text.translate(str.maketrans('', '', string.punctuation))
    # remove digits
    text = re.sub(r'\d+', ' ', text)
    # collapse multiple spaces
    text = re.sub(r'\s+', ' ', text).strip()
    return text
```

#apply text cleaning

```python
#  6: Apply cleaning to both datasets
df_flipkart_data['cleaned_review'] = df_flipkart_data['review_text'].apply(clean_text)
df_amazon_data['cleaned_review']  = df_amazon_data['review_text'].apply(clean_text)

display("Example Flipkart cleaned review:\n", df_flipkart_data['cleaned_review'].iloc[0] )#cleaned the text column, select from 1st row
display("\nExample Amazon cleaned review:\n", df_amazon_data['cleaned_review'].iloc[0])
```

#Keep Only Needed Columns & Merge

```python
#7: Select consistent columns and merge
use_cols = ['review_text', 'rating', 'label', 'platform', 'cleaned_review']

df_flipkart_use = df_flipkart_data[use_cols].copy()
df_amazon_use  = df_amazon_data[use_cols].copy()

df_all_data = pd.concat([df_flipkart_use, df_amazon_use], axis=0, ignore_index=True) # each row removes, # reset the index

display("Merged shape:", df_all_data.shape)
display("\nPlatform counts:\n", df_all_data['platform'].value_counts())
display("\nLabel counts:\n", df_all_data['label'].value_counts())
```

#Remove Missing & Duplicates

```python
# 8: Drop the remaining missing values
df_all_data = df_all_data.dropna(subset=['cleaned_review', 'label', 'rating'])
df_all_data.reset_index(drop=True, inplace=True)#new row numbers  ,#remove old index , #change original data

# Drop duplicates
df_all_data = df_all_data.drop_duplicates(subset=['cleaned_review'])
df_all_data.reset_index(drop=True, inplace=True) #new row numbers  ,#remove old index , #change original data

display("After cleaning, shape:", df_all_data.shape)
display("Nulls:\n", df_all_data.isnull().sum())
```

#Prepare Data for Modelling

```python
#11: Prepare X, y and train/test split

X = df_all_data['cleaned_review']
y = df_all_data['label']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,  # 20% for testing
    random_state=42, # control randomness in traing/testing
    stratify=y  # keeps class balance

)

print("Train size:", X_train.shape[0])
print("Test size :", X_test.shape[0])
```

#TF-IDF Vectorisation

```python
#  12: TF-IDF vectorisation
tfidf = TfidfVectorizer(
    max_features=5000,  ## limit vocabulary size
    stop_words='english', ##removes common english stop words
    ngram_range=(1,2) # use single words and two-word phrases
)

## use single words and two-word phrases
Xtr = tfidf.fit_transform(X_train)

# Transform test data using same vocabulary
Xte = tfidf.transform(X_test)

# Check data dimensions
print("Xtr shape:", Xtr.shape)
display("Xte shape:", Xte.shape)
```

#Logistic Regression Model

```python
# 13: Logistic Regression baseline
lr = LogisticRegression(max_iter=1000) # increase iterations for convergence

#train the model on training data
lr.fit(Xtr, y_train)

# predict on test data
y_pred_lr = lr.predict(Xte)

#check accuracy
print("Logistic Regression Accuracy:", accuracy_score(y_test, y_pred_lr))
print("\nClassification Report (Logistic Regression):\n")
print(classification_report(y_test, y_pred_lr))
```

#SVM Model (LinearSVC)

```python
#  14: SVM (LinearSVC)
svm = LinearSVC()  # linear kernal for text data

#train the model
svm.fit(Xtr, y_train)

#predict on test data
y_pred_svm = svm.predict(Xte)

#check accuracy
print("SVM Accuracy:", accuracy_score(y_test, y_pred_svm))
print("\nClassification Report (SVM):\n")
print(classification_report(y_test, y_pred_svm))
```

#Simple Hyperparameter Tuning for SVM (C value)

```python
# 15: Hyperparameter tuning for SVM C

#difine values of c to try (controls regulations strength)
param_grid = {
    'C': [0.1, 1, 5]} # c=simple model,C=complex model

#base linear svm model for tuning
svm_base = LinearSVC()

#grid search with cross-validation
grid_svm = GridSearchCV(
    estimator=svm_base, #model to tune
    param_grid=param_grid,#parameter values to test
    scoring='f1_macro',#suitable for imbalanced multi-class
    cv=3, #3-fold validation
    n_jobs=-1 #use all cpu cores
)

#train grid search on training data
grid_svm.fit(Xtr, y_train)

#display best parameter found
print("Best params:", grid_svm.best_params_)

#extract best tuned svm model
best_svm = grid_svm.best_estimator_

#predict sentiment using tuned model
y_pred_best_svm = best_svm.predict(Xte)

#evaluate tuned model performance
print("\nTuned SVM Accuracy:", accuracy_score(y_test, y_pred_best_svm))
print("\nClassification Report (Tuned SVM):\n")
print(classification_report(y_test, y_pred_best_svm))
```
